```python
'''
資料科學的研究流程(循環、非線性):
資料載入。
資料清理。
資料轉換。
資料視覺化。
模型。
溝通。
'''
```

```python
'''
機器學習的定義:三個要素、一個但書
[要素]
experience E (訓練資料/歷史資料)
tasks T (目的性)
performance measure P (準確度)
[但書]
if its performance at tasks in T, as measured by P,
improves with experience E.
(它在任務T中的表現(由指標P來衡量)，會隨著經驗E的增加而提升。)
'''
```

```python
'''
機器學習的模型可在分為:
[監督式學習]:訓練資料中具備以實現的數值或標籤。
迴歸:數值預測的任務。
分類:類別預測的任務。
[非監督式學習]:訓練資料中「不」具備任何的數值或標籤。
'''
```

```python
'''
模型(Model)是一個類似於「函數」的概念，由參數與運算組成。
模型的參數以及運算可以透過不同的方式生成，生成方式包含規則敘述與歷史資料訓練。

透過規則敘述生成參數以及運算，稱為基於規則模型(Rule-based medel)或稱專家模型。
透過歷史資料訓練生成參數以及運算，稱為基於演算法的模型(Algorithm-based medel)或稱基於機器學習的模型。

基於規則模型:問題能用人類語言描述邏輯、撰寫規則。答案不能容忍誤差。給定整數判斷它是否為奇數、偶數或質數。
基於演算法的模型:問題非領域專家不容易描述邏輯、撰寫規則。答案能夠容忍誤差。給定圖片判斷是貓或狗、給定房屋資訊預測房價。
'''
```

```python
'''
關於訓練、驗證與測試:
機器學習專案中的訓練、驗證來自具備已實現數值或標籤資料集。
測試則來自未實現數值或標籤資料集。
在從未見過的測試資料上之表現將決定他是否會被部署到正式環境開始運作。
'''
```

```python
'''
減少訓練誤差的具體做法:
增加梯度遞減的訓練次數。
透過交叉驗證(Cross validation)的技巧消弭訓練與驗證資料切割所造成的誤差。
增加特徵矩陣的欄位。

減少訓練誤差與測試誤差的間距:
引用正規化（Regularization）的技巧。
減少訓練誤差的過程，很有可能伴隨而來的是驗證或測試誤差的升高。
模型對於訓練資料過於熟悉，而降低了它的泛化能力。
這樣的狀態我們稱之為「過度配適」（Overfitting）。
'''
```

```python
'''
Scikit-learn 是 Python 機器學習的第三方模組，
透過它可以進行監督式以及非監督式學習，提供了模型訓練、資料預處理、模型選擇以及模型評估等功能。

Scikit-Learn 應用程式介面設計原則:
一致性:每個轉換器類別都有 fit_transform() 方法。
可檢查性:每個轉換器或預測器都有屬性讓使用者檢視轉換或預測的規則。
不擴增新類別。
可組合性。
合理預設參數:每個轉換器、預測器都可以用預設參數建立物件。
'''
```

```python
'''
主成分分析(Principal Component Analysis, PCA):為一種常用的降低數據維度(Dimensionality reduction)的手法。
主要的核心概念是用最少的維度來呈現原本高維度數據的訊息:
將原本p維的特徵(features)以k維的線性組合來表達(k < p)，
藉由找到的k維新特徵，能夠捕捉原本p維數據中的大部分特性來解釋數據。
換言之，希望能夠在盡量保有原始數據訊息的情況下，以更精簡化的形式來呈現原始數據。
'''
```

```python
'''
主成分分析的步驟:
1.對數據做標準化。
2.建立共變異數矩陣。
3.分解共變異數矩陣為特徵向量及特徵值。
4.排序特徵值並且以該順序排列特徵向量。
5.映射數據至主成分。
'''
```

```python
import numpy as np

# 對數據做標準化
# 可以用 StandardScaler 取代
def standardize(X):
    mean_X = np.mean(X, axis = 0)
    std_X = np.std(X, axis = 0)
    return (X - mean_X) / std_X


# 建立共變異數矩陣
# 可以用 np.cov() 取代
def covariance_matrix(X):
    m = X.shape[0]
    X_T = np.transpose(X)
    X_T_dot_X = X_T.dot(X) # (n, m) dot (m, n)
    return X_T_dot_X / (m - 1)


# 分解共變異數矩陣為特徵向量及特徵值
from numpy.linalg import eigh

def get_eig_vals_vecs(X):
    X_standard = standardize(X)
    cov_mat = covariance_matrix(X_standard)
    eig_vals, eig_vecs = eigh(cov_mat)
    return eig_vals, eig_vecs # 回傳特徵值(Eigenvalue)與特徵向量(Eigenvector)。


# 排序特徵值並且以該順序排列特徵向量
def get_principal_components(eigen_vals, eigen_vecs, k=2): # 降維要降到2維自己決定，k=2。
    sort_index = np.argsort(eigen_vals)[::-1]
    principal_components = eigen_vecs[:,sort_index]
    return principal_components[:, 0:k]


# 映射數據至主成分
def project_with_principal_components(X, principal_components):
    X_proj = X.dot(principal_components)
    return X_proj
```

```python
X = player_stats[["heightMeters", "weightKilograms", "ppg", "rpg", "apg", "bpg"]].values
print(X.shape)

eig_vals, eig_vecs = get_eig_vals_vecs(X)
print(eig_vals)
print(eig_vecs)

principal_components = get_principal_components(eig_vals, eig_vecs)
print(principal_components)
```

```python
project_X = project_with_principal_components(standardize(X), principal_components)
project_X.shape
```

```python
# 與 Scikit-Learn 的 PCA 類別對照
from sklearn.decomposition import PCA

X = player_stats[["heightMeters", "weightKilograms", "ppg", "rpg", "apg", "bpg", "mpg", "spg"]]
standardize_X = standardize(X)
pca = PCA(n_components=2)
pca.fit(standardize_X)
print(pca.components_.T)
```
