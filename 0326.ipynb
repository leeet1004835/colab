```python
'''
如果一個適合跑馬拉松的天氣定義為乾且冷，
假設比賽日乾燥與低溫的機率均為50%，
適合跑馬拉松的機率為25%。
'''
def is_a_good_marathon_weather(is_dry: bool, is_cold: bool) -> bool:
    is_good = (is_dry and is_cold) or (not is_dry and not is_cold)
    return is_good

print(is_a_good_marathon_weather(True, True))
print(is_a_good_marathon_weather(True, False))
print(is_a_good_marathon_weather(False, True))
print(is_a_good_marathon_weather(False, False))
```

```python
'''
list 採用兩個方向的索引機制
由左至右:從0開始
由右至左:從-1開始
'''
primes = [2, 3, 5, 7, 11]

print(primes[0]) # 2
print(primes[1]) # 3
print(primes[2]) # 5
print(primes[3]) # 7
print(primes[4]) # 11

print(primes[-1]) # 11
print(primes[-2]) # 7
print(primes[-3]) # 5
print(primes[-4]) # 3
print(primes[-5]) # 2
```

```python
'''
slicing[start:stop:step]
start 起始位置(包含)預設為0
stop 終止位置(排除)預設為list的長度
step 間隔預設為1
'''
print(primes[0:3:1]) # 前三個
print(primes[:3]) # 前三個

print(primes[2:5:1]) # 後三個
print(primes[2:]) # 後三個

print(primes[0:5:2]) # 每隔一個取一次
print(primes[0::2]) # 每隔一個取一次
print(primes[::2]) # 每隔一個取一次

# 更新list中的資料值
print(primes)
primes[-1] = 13
print(primes)
```

```python
'''
list 支援的運算符
+運算符:連接lists
*運算符:複製list中的元素
'''
print(primes + [15, 17, 19, 21])
print(primes * 3)
```

```python
# set是一種「無序」、儲存「獨一值」並且能夠進行「集合運算」的資料結構。
primes = {2, 3, 5, 7, 7, 7, 7, 5, 5, 5, 11, 13, 17, 19} # 7 is duplicated
odds = {1, 3, 5, 7, 9, 9, 9, 9, 5, 5, 11, 13, 15, 17, 19} # 9 is duplicated
print(primes)
print(odds)
```

```python
# set支援集合運算
# 交集
primes.intersection(odds) # 既是奇數、又是質數
```

```python
# set支援集合運算
# 聯集
primes.union(odds) # 是奇數或是質數
```

```python
# set支援集合運算
# 差集
print(primes.difference(odds)) # 是質數但不是奇數
print(odds.difference(primes)) # 是奇數但不是質數
print(primes.symmetric_difference(odds)) # [是質數但不是奇數] 或 [是奇數但不是質數]
```

# 使用「條件」與「縮排」建立條件敘述

```python
# 使用if依據條件決定是否執行程式區塊
x = 56
if x > 0: # 56 > 0 -> True
    print(f"{x} is positive.") # Run!
```

```python
# 使用if依據條件決定是否執行程式區塊
x = 0
if x > 0: # 0 > 0 -> False
    print(f"{x} is positive.") # Not executed!
```

```python
# 使用if依據條件決定是否執行程式區塊
def return_if_positive(x: int):
    if x > 0:
        return f"{x} is positive."

print(return_if_positive(56))
```

```python
# 使用if...else...依據條件決定執行兩個程式區塊其中的一個
def return_whether_positive_or_not(x: int):
    if x > 0:
        return f"{x} is positive." # x > 0 is True
    else:
        return f"{x} is not positive." # x > 0 is False

print(return_whether_positive_or_not(56))
print(return_whether_positive_or_not(0))
print(return_whether_positive_or_not(-56))
```

```python
# 使用if...elif...else...依據條件決定執行多個程式區塊其中的一個
def return_positive_neutral_negative(x: int):
    if x > 0:
        return f"{x} is positive." # x > 0 is True
    elif x == 0:
        return f"{x} is neutral." # x == 0 is True
    elif x < 0:
        return f"{x} is not negative." # x < 0 is True

print(return_positive_neutral_negative(56))
print(return_positive_neutral_negative(0))
print(return_positive_neutral_negative(-56))
```
