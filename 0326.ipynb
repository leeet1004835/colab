```python
'''
將華氏溫度轉換為攝氏溫度的自行定義函數
'''
def convert_fahrenheit_to_celsius(f: int) ->float:
    c = (f - 32) * 5 / 9
    return c
```

```python
print(convert_fahrenheit_to_celsius(212))
print(convert_fahrenheit_to_celsius(32))
```

```python
def hello_to_anyone(anyone):
    output = "Hello" + " " + anyone
    return output

print(hello_to_anyone("Tony"))
```

```python
'''
透過str.format()方法以及f-string語法搭配大括號{}指定顯示格式
"{}".format(object_name) # str.format() 方法
f"{object_name}" # f-string 語法
'''
def hello_to_anyone(anyone: str, another_one: str) -> str:
    #output = "Hello {}, {}!".format(anyone, another_one) # {} as a placeholder
    output = f"Hello {anyone}, {another_one}!" # f-string
    return output

print(hello_to_anyone("Python", "Rust"))
```

```python
# 指定小數點n位的浮點數格式:.nf
math_e = 2.71828
print("Math e:{}".format(math_e))
print("Math e:{:.1f}".format(math_e)) # 得到一位小數
print("Math e:{:.2f}".format(math_e)) # 得到兩位小數
print("Math e:{:.3f}".format(math_e)) # 得到三位小數
print("Math e:{:.4f}".format(math_e)) # 得到四位小數
```

```python
# 指定小數點n位的浮點數格式:.nf
math_e = 2.71828
print(f"Math e:{math_e}")
print(f"Math e:{math_e:.1f}") # 得到一位小數
print(f"Math e:{math_e:.2f}") # 得到兩位小數
print(f"Math e:{math_e:.3f}") # 得到三位小數
print(f"Math e:{math_e:.4f}") # 得到四位小數
```
