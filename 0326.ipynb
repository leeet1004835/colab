```python
'''
將華氏溫度轉換為攝氏溫度的自行定義函數
'''
def convert_fahrenheit_to_celsius(f: int) ->float:
    c = (f - 32) * 5 / 9
    return c
```

```python
print(convert_fahrenheit_to_celsius(212))
print(convert_fahrenheit_to_celsius(32))
```

```python
'''
使用成對的單引號'、雙引號"或三個雙引號"""形成str
'''
hello_world_1 = 'Hello World!' # single quotes
hello_world_2 = "Hello World!" # double quotes
hello_world_3 = """Hello World!""" # triple quotes
print(type(hello_world_1))
print(type(hello_world_2))
print(type(hello_world_3))
```
