```python
'''
將華氏溫度轉換為攝氏溫度的自行定義函數
'''
def convert_fahrenheit_to_celsius(f: int) ->float:
    c = (f - 32) * 5 / 9
    return c
```

```python
'''
透過str.format()方法以及f-string語法搭配大括號{}指定顯示格式
"{}".format(object_name) # str.format() 方法
f"{object_name}" # f-string 語法
'''
def hello_to_anyone(anyone: str, another_one: str) -> str:
    #output = "Hello {}, {}!".format(anyone, another_one) # {} as a placeholder
    output = f"Hello {anyone}, {another_one}!" # f-string
    return output

print(hello_to_anyone("Python", "Rust"))
```

```python
'''
關係運算符
等於 ==
不等於 !=
大於 > 大於等於 >=
小於 < 小於等於 <=
包含於 in
不包含於 not in
'''
my_lucky_number = 5566
my_favorite_number = "5566"
print(my_lucky_number == 5566)
print(my_lucky_number != 5566)
print(my_lucky_number > 5566)
print(my_lucky_number >= 5566)
print(my_lucky_number < 5566)
print(my_lucky_number <= 5566)
print('56' in my_favorite_number)
print('78' not in my_favorite_number)
```

```python
'''
邏輯運算符
交集 and
聯集 or
非 not
'''
bool_false = False
bool_true = True
print(not bool_false)
print(not bool_true)
```

```python
# 交集and運算符兩側必須都是True判斷結果才會是True
print(bool_true and bool_true)
print(bool_false and bool_true)
print(bool_true and bool_false)
print(bool_false and bool_false)

# 聯集or運算符兩側必須都是False判斷結果才會是False
print(bool_false or bool_false)
print(bool_false or bool_true)
print(bool_true or bool_false)
print(bool_true or bool_true)
```

```python
'''
如果一個適合跑馬拉松的天氣定義為乾且冷，
假設比賽日乾燥與低溫的機率均為50%，
適合跑馬拉松的機率為25%。
'''
def is_a_good_marathon_weather(is_dry: bool, is_cold: bool) -> bool:
    is_good = (is_dry and is_cold) or (not is_dry and not is_cold)
    return is_good

print(is_a_good_marathon_weather(True, True))
print(is_a_good_marathon_weather(True, False))
print(is_a_good_marathon_weather(False, True))
print(is_a_good_marathon_weather(False, False))
```
