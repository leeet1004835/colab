```python
# 從list創造ndarray
prime_list = [2, 3, 5, 7, 11]
print(prime_list)
print(type(prime_list))

prime_array = np.array(prime_list)
print(prime_array)
print(type(prime_array))
```

```python
print(prime_list + prime_list)
print(prime_list * 3)

print(prime_array + prime_array)
print(prime_array * 3)
```
