```python
'''
如何寫作一個while迴圈:
在迴圈程式區塊之前定義一個物件設定起始值。
設計條件讓程式區塊重複執行的次數符合我們的需求。
記得在程式區塊中更新物件的值。
'''
```

```python
'''
while 迴圈:印出5次"Hello world!"
'''
i = 0 # start
while i <= 4: # stop
  print("Hello world!")
  i += 1 # step

# i = 0 # zero-based indexing
# while i <= 4: # condition
#   print("Hello world!")
#   i += 1
```
