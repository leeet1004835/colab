```python
'''
如何寫作一個while迴圈:
在迴圈程式區塊之前定義一個物件設定起始值。
設計條件讓程式區塊重複執行的次數符合我們的需求。
記得在程式區塊中更新物件的值。
'''
```

```python
'''
while 迴圈:印出5次"Hello world!"
'''
i = 0 # start
while i <= 4: # stop
  print("Hello world!")
  i += 1 # step: i = i + 1

# i = 0 # zero-based indexing
# while i <= 4: # condition
#   print("Hello world!")
#   i += 1 # compound operators
```

```python
'''
使用for走訪可迭代類別中的所有元素
for 元素 in 可迭代類別:
  程式區塊
'''
# iterable可迭代類別
str_as_iterable = "Luke Skywalker"
# iter(str_as_iterable)
for char in str_as_iterable:
    print(char)
```

```python
'''
如何寫作一個for迴圈
建立一個可迭代類別。
可迭代類別如果是數列，可透過內建函數range()建立。
'''

data_structure_as_iterable = [1, 2, 3, 4, 5]
for item in data_structure_as_iterable:
    print(item)
```

```python
'''
range()函數有三個參數可以設定數列內容
start 數列的起始整數，即第0個整數(包含)，預設值為0。
stop 數列的終止整數，即第-1個整數(排除)。
step 數列的公差，預設值為1。
'''
range(10)
print(list(range(10)))
print(list(range(0, 10, 1)))
print(list(range(1, 11, 1)))
```

```python
# 運用 + 運算符連lists
vowels = {
    "upper_cased": ("A", "E", "I", "O", "U"),
    "lower_cased": ("a", "e", "i", "o", "u")
}
# flat_vowels = ["a","e","i","o","u","A","E","I","O","U"]

# flat_vowels = vowels["upper_cased"] + vowels["lower_cased"]
# print(flat_vowels)

flat_vowels = [] # list
for val in vowels.values():
    flat_vowels += list(val)
print(flat_vowels)
```
