```python
'''
while 迴圈:印出5次"Hello world!"
'''
i = 0 # start
while i <= 4: # stop
  print("Hello world!")
  i += 1 # step: i = i + 1

# i = 0 # zero-based indexing
# while i <= 4: # condition
#   print("Hello world!")
#   i += 1 # compound operators
```

```python
'''
如何寫作一個for迴圈
建立一個可迭代類別。
可迭代類別如果是數列，可透過內建函數range()建立。
'''

data_structure_as_iterable = [1, 2, 3, 4, 5]
for item in data_structure_as_iterable:
    print(item)
```

```python
'''
range()函數有三個參數可以設定數列內容
start 數列的起始整數，即第0個整數(包含)，預設值為0。
stop 數列的終止整數，即第-1個整數(排除)。
step 數列的公差，預設值為1。
'''
range(10)
print(list(range(10)))
print(list(range(0, 10, 1)))
print(list(range(1, 11, 1)))
```

```python
days_of_week = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

days_of_week_dict = {} # 先建立一個空字典

for dow in days_of_week:
    acronym = dow[:3] # [start:stop:step]
    a_u = acronym.upper()
    days_of_week_dict[a_u] = dow

print(days_of_week_dict)
```

```python
# 將print移入迴圈內
for dow in days_of_week:
  if dow == "Thursday":
    break
  print(dow) # 會印出 Mon, Tue, Wed
```

```python
# 將print移入迴圈內
for dow in days_of_week:
  if dow == "Thursday":
    continue # 跳過本次，繼續下一個
  print(dow) # 會印出 Mon, Tue, Wed, Fri
```

```python
def get_first_and_last_character(x):
  return x[0], x[-1]

get_first_and_last_character("Luke Skywalker")
```

```python
def get_capital_char(x):
  x_split = x.split()
  captial_char = list()
  for char in x_split:
    captial_char.append(char[0])
  return captial_char

get_capital_char("Huntrix girls to the world!")
```

```python
def sum_and_square(x):
  sum_x = sum(x)
  square_x = sum_x**2
  return square_x

sum_and_square((2, 3, 5)) # 多一層小括號
```

```python
# 利用*標註彈性參數

# def sum_and_square(x1, x2, x3, ..., xn):
  # sum_x = sum((x1, x2, x3, ..., xn))

def sum_and_square(*args):
  sum_x = sum(args)
  square_x = sum_x**2
  return square_x
  # return args

sum_and_square(2, 3) # 少了一層小括號
```

```python
# 用**標註具有「鍵」與「值」的彈性參數
def print_country_and_capital(**kwargs):
    return kwargs

print_country_and_capital(TWN = "Taipei")
```

```python
def print_country_and_capital(**kwargs):
    for k, v in kwargs.items():
        print(f"Country: {k}; Capital: {v}")

print_country_and_capital(TWN = "Taipei")
```

```python
print_country_and_capital(TWN = "Taipei", USA = "Washington D.C.", JPN = "Tokyo")
```
