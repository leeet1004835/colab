```python
import numpy as np
```

```python
# 從 list 創造 ndarray
prime_list = [2, 3, 5, 7, 11]
print(prime_list)
print(type(prime_list))

prime_array = np.array(prime_list)
print(prime_array)
print(type(prime_array))
```

```python
print(prime_list + prime_list) # concatenation
print(prime_list * 3) # duplicate/extend

print(prime_array + prime_array) # element-wise add
print(prime_array * 3) # element-wise multiply
print(prime_array ** 2) # element-wise power
```

```python
'''
利用 NumPy 函數創造內容元素相同的 ndarray
np.zeros()
np.ones()
np.full()
'''
print(np.zeros(5, dtype = int)) # dtype = data type
print(np.ones(5, dtype = int))
print(np.ones(10, dtype = int) * 5566)
print(np.full(10, 5566, dtype = int))
```

```python
'''
利用 NumPy 函數創造數列型態的 ndarray
np.arange(start, stop, step)
np.linspace(start, stop, num)值得注意的是stop參數預設為包含。
'''
print(np.arange(1, 11, 2))
print(np.linspace(1, 100, 10, dtype = int)) # 固定間距的數列
```

```python
# 可內積(dot)的
# (m, n) dot (i, j), n == i
arr_1 = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])
arr_2 = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])
print(arr_1.shape)
print(arr_2.T.shape)
np.dot(arr_1, arr_2.T)
print(len(arr_1.shape))
print(arr_1.ndim)
```

```python
'''
常用的 ndarray 屬性:
ndarray.ndim 維度數。
ndarray.shape 外型。
ndarray.size 元素個數。
ndarray.dtype 資料類別。
'''
prime_list = [2, 3, 5, 7, 11]
prime_array = np.array(prime_list)
print(prime_array)
print(prime_array.ndim)
print(prime_array.shape)
print(prime_array.size)
print(prime_array.dtype)
```

```python
'''
零維 ndarray: 純量
一維 ndarray: 向量
二維 ndarray: 矩陣
三維或者n維 ndarray: 張量
'''
an_integer = 5566
print(an_integer)

scalar = np.array(an_integer)
print(scalar)
print(scalar.shape)
print(scalar.ndim)

vector = np.array([5, 5, 6, 6])
print(vector)
print(vector.shape) # 四個元素
print(vector.ndim) # 維度是1

matrix = np.array([
    [5, 5],
    [6, 6]
])
print(matrix)
print(matrix.shape) # (m, n)，2 rows, 4 columns
print(matrix.ndim) # 維度是2

tensor = np.array([
    [
        [5, 5],
        [6, 6]
    ],
    [
        [5, 5],
        [6, 6]
    ],
    [
        [5, 5],
        [6, 6]
    ]
])
print(tensor)
print(tensor.shape) # (l, m, n)
print(tensor.ndim) # 維度是3
```

```python
print(prime_list)
print(prime_array)

print(prime_array[0])
print(prime_array[1])
print(prime_array[2])
print(prime_array[3])
print(prime_array[4])

print(prime_array[-1])
print(prime_array[-2])
print(prime_array[-3])
print(prime_array[-4])
print(prime_array[-5])

print(prime_array[::])
print(prime_array[:2])
print(prime_array[2:])
```

```python
# ndarray 與 list 相異的地方
matrix_by_list = [[5, 5],
                  [6, 6],
                  [55, 66]]
matrix = np.array([[5, 5],
                  [6, 6],
                  [55, 66]])
print(matrix_by_list)
print(matrix)

print(matrix_by_list[2][1])
print(matrix[2, 1]) # 取第二列第一欄

print(matrix_by_list[0][1], matrix_by_list[1][1], matrix_by_list[2][1])
print(matrix[:, 1]) # 取所有列的第一欄
print(matrix[:, [1]]) # 取所有列的第一欄，但不降維
```

```python
# 同質性資料結構類別，若資料維不同型別，會進行資料型別同質性轉換
mixed_type_list = [False, True, 55, 66, 3.14, "Hello world!"]
homogeneous_array = np.array([False, True, 55, 66, 3.14, "Hello world!"])
for elem in mixed_type_list:
  print(type(elem))
```

```python
print(homogeneous_array)
for elem in homogeneous_array:
  print(type(elem))
```

```python
'''
支援特殊的 indexing 語法
Fancy indexing
Boolean indexing
'''
print(prime_list)
print(prime_list[::2])

print(prime_array)
print(prime_array[[0, 1, -1]]) # Fancy indexing，可以更有彈性地取出 ndarray 元素
print(prime_array[[True, True, False, False, True]]) # Boolean indexing
```

```python
interger_array = np.arange(1, 100)
print(interger_array)

interger_array[interger_array % 2 == 0] # 等同於下面兩行程式
# bool_indexing = interger_array % 2 == 0
# interger_array[bool_indexing]
```

```python
'''
常用的 ndarray 操作:
調整外型。
複製。
合併。
分割。
'''
print(matrix)
print(matrix.ravel()) # 直接回到1維
print(tensor)
print(tensor.ravel()) # 直接回到1維
```

```python
arr = np.arange(100)

print(arr.shape)
print(arr.reshape(25, 4))
print(arr.reshape(5, 5, 4))
```

```python
print(arr.reshape(5, -1)) # 在其他維度已經決定時可以方便地指定 -1 給最後一個維度
print(arr.reshape(-1, 5))
```

```python
arr = np.arange(10)
mat = arr.reshape(-1, 2) # mat is a view of arr
print(arr)
print(mat)
```

```python
arr[-1] = 999
print(arr)
print(mat)
```

```python
'''
合併
使用 np.concatenate() 函數合併
指定參數 axis = 0 垂直合併 (預設值)
指定參數 axis = 1 水平合併
'''
x1 = np.arange(1, 11).reshape(-1, 1)
print(x1)
x0 = np.ones(x1.size).reshape(-1, 1)
print(x0)

# y = error(residual) + w1x1 + ... + wnxn
# y = error*1 + w1*x1 + ... + wn*xn
# y = w0*x0 + w1*x1 + ... + wn*xn

x = np.concatenate((x0, x1), axis = 1) # 水平合併
print(x)
```

```python
'''
分割
使用 np.split()函數分割
指定參數 axis = 0 垂直分割 (預設值)
指定參數 axis = 1 水平分割
'''
x
np.split(x, [2], axis = 0)
```

```python
array_range = np.arange(20).reshape(-1, 2)
print(array_range)
upper_array, lower_array = np.split(array_range, [2], axis = 0)
print(upper_array)
print(lower_array)
```

```python
valid_percentage = 0.2
valid_index = int(x.shape[0] * valid_percentage)
print(valid_index)

train_percentage = 0.7
train_index = int(x.shape[0] * train_percentage)
print(train_index)

valid_set, train_set, test_set = np.split(np.arange(x.shape[0]), [valid_index, train_index])
print(valid_set)
print(train_set)
print(test_set)
```
